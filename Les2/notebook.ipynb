## Over de zelfcontroles

Na sommige cellen staat een zelfcontrole. Die kijkt naar je resultaat en drukt af
wat ze ziet. Het is een controle, geen punt. Ze stopt je notebook nooit, dus je
kan altijd verder.

Heb je het anders opgelost en klaagt de controle over een naam? Van een aantal
namen zegt de opgave dat ze vastliggen, want onze code gebruikt ze verderop. Alle
andere namen kies je zelf. Hernoem dan gerust, of lees gewoon verder.

Voer de cel hieronder één keer uit. Die zet de hulpjes voor de controles klaar.

```python
# Hulpjes voor de zelfcontroles. Je hoeft deze cel niet te lezen.

def waarde_van(naam):
    """De waarde met die naam, of None als die naam er nog niet is."""
    waarde = globals().get(naam, None)
    if waarde is Ellipsis:
        return None
    return waarde


def als_lijst(waarde):
    """Maakt er een gewone lijst van, of None als het geen lijst is."""
    if waarde is None or isinstance(waarde, str):
        return None
    try:
        return list(waarde)
    except TypeError:
        return None


def plat(waarde):
    """Zoals als_lijst, maar haalt lijstjes in de lijst ook uit elkaar."""
    lijst = als_lijst(waarde)
    if lijst is None:
        return None
    uit = []
    for stuk in lijst:
        binnenin = None if isinstance(stuk, str) else als_lijst(stuk)
        if binnenin is None:
            uit.append(stuk)
        else:
            uit.extend(binnenin)
    return uit


def als_getal(waarde):
    """Maakt er een kommagetal van, of None als dat niet lukt."""
    try:
        return float(waarde)
    except (TypeError, ValueError):
        return None


def zelfcontrole(titel, punten):
    """Drukt af wat nog niet klopt. Deze controle stopt je notebook nooit."""
    open_punten = []
    for klopt, tekst in punten:
        if not klopt:
            open_punten.append(tekst)
    if not open_punten:
        print(titel + ": in orde.")
        return
    print(titel + ": dit valt ons op.")
    for tekst in open_punten:
        print("  - " + tekst)


print("De zelfcontroles staan klaar.")
```

## Wat betekenen die scores?

> Slide: Oplossing Stap 14

Een score van 1 betekent dat het model perfect voorspelt. Bij een score van 0 is
het model niet beter dan een gemiddelde gok. Zakt de score onder 0? Dan doet het
model het nog slechter dan die gok.

Voer de twee cellen van stap 14 een paar keer opnieuw uit. De train set en de
