`print()` toont de tekst zoals ze eruitziet. `repr()` toont ook wat je niet ziet:
het regeleinde, spaties op het einde, en of iets tekst is of een getal.

Voer de cel hieronder uit en vergelijk de twee regels.

```python
print('12.4\n')
print(repr('12.4\n'))
```

De eerste regel laat een lege rij achter en je ziet niet waarom. De tweede toont
`'12.4\n'`. Nu zie je het regeleinde staan, en de aanhalingstekens vertellen je dat
dit tekst is en geen getal.

`repr()` naar zodra iets niet klopt.
