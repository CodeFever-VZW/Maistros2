## Wat betekenen die scores?

> Slide: Oplossing Stap 14

Een score van 1 betekent dat het model perfect voorspelt. Bij een score van 0 is
het model niet beter dan een gemiddelde gok. Zakt de score onder 0? Dan doet het
model het nog slechter dan die gok.

Voer de twee cellen van stap 14 een paar keer opnieuw uit. De train set en de
test set worden elke keer willekeurig gekozen. Daardoor krijg je elke keer andere
scores.

Bij de mannen springt de score het hardst op en neer. Er zitten veel minder
mannen in de data. Die score kan zelfs onder 0 zakken.
