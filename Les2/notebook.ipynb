## Tussendoor 3 - herhaling dictionaries

> Slide: Tussendoor 3: herhaling dictionaries

Hier programmeer je niets. Voer de cel uit.

```python
leeftijden = {}
leeftijden["Tessa"] = 24
leeftijden["Peter"] = 52
leeftijden["Blahaj"] = 9
print(leeftijden["Blahaj"])

# Een value aanpassen.
leeftijden["Blahaj"] = leeftijden["Blahaj"] + 1
print(leeftijden)

# Testen of een key erin zit.
if "Albert" in leeftijden:
    print("Albert is " + str(leeftijden["Albert"]) + " jaar.")
else:
    print("Nog nooit van Albert gehoord...")

# Alle keys overlopen.
for naam in leeftijden:
    print(naam, leeftijden[naam])
```
