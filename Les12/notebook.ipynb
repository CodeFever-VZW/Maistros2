## Stap 1: speel het spel zelf

> Slide: Stap 1: speel het spel zelf

```python
from gymnasium.utils.play import play
import gymnasium as gym
import pygame

# Welke toets hoort bij welke actie: pijltjestoetsen of WASD
keys_to_action = {
    (pygame.K_UP,): 0,    (pygame.K_w,): 0,    # omhoog
    (pygame.K_RIGHT,): 1, (pygame.K_d,): 1,    # rechts
    (pygame.K_DOWN,): 2,  (pygame.K_s,): 2,    # omlaag
    (pygame.K_LEFT,): 3,  (pygame.K_a,): 3,    # links
}
# Maak de omgeving aan met visualisatie
omgeving = gym.make("CliffWalking-v0", render_mode="rgb_array")

print("🎮 Speel het spel met WASD of de pijltjestoetsen! 👉 Druk ESC om te stoppen.")
print("🎯 Doel: kom van linksonder naar rechtsonder zonder in de afgrond te vallen!\n")
play(omgeving, keys_to_action=keys_to_action, wait_on_player=True)
```
