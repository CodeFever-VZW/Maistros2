## Oefening 2 - stap 2: de functies voor een frequentietabel klaarzetten

> Slide: Oefening 2: Hotelreservaties - Stap 2

```python
def frequentieTabel(lijst):
    tabel = {}
    for element in lijst:
        if not element in tabel:
            tabel[element] = 1
        else:
            tabel[element] = tabel[element] + 1
    return tabel


def printFrequentieTabel(ftabel, titel):
    print(titel)
    sorted_dict = sorted(ftabel.items(), key=lambda x: x[1], reverse=True)
    for key, value in sorted_dict:
            print("  " + str(key) + ": " + str(value))
```
