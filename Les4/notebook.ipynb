## Stap 6 - accuracy

> Slide: MNIST - Stap 6

```python
# accuracy berekenen
def accuracy(voorspellingen,antwoorden):
    nr_correct = 0
    for antwoord,voorspelling in zip(antwoorden,voorspellingen):
        if antwoord == voorspelling:
            nr_correct += 1
    return nr_correct/len(voorspellingen)

# JOUW CODE (6)
...
```
