## Eerst zelf spelen: het gemakkelijkere spelletje klaarzetten

> Slide: requirements.txt

> Slide: We proberen het eerst zelf te spelen (code 1 van 2)

```python
import math
from gymnasium.envs.classic_control.cartpole import CartPoleEnv
from gymnasium.envs.registration import register

class EasyCartPole(CartPoleEnv):
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
        self.masspole = 0.02
        self.length = 0.75
        self.tau = 0.01
        self.theta_threshold_radians = 45 * 2 * math.pi / 360  # ongeveer 45 graden

# Registreer het gemakkelijkere spelletje als een nieuwe omgeving
register(id="EasyCartPole-v1", entry_point=EasyCartPole)
```

## Eerst zelf spelen: het spelletje starten

> Slide: We proberen het eerst zelf te spelen (code 2 van 2)

```python
import gymnasium as gym
import pygame
from gymnasium.utils.play import play
keys_to_action = {
    (pygame.K_LEFT,): 0,   # Pijltje links
    (pygame.K_RIGHT,): 1   # Pijltje rechts
}
env = gym.make("EasyCartPole-v1", render_mode="rgb_array")
play(env, keys_to_action=keys_to_action, wait_on_player=True)
```
