## Pinguïns Classificeren - Stap 2: de data opsplitsen

```python
from sklearn.model_selection import train_test_split

X_train, X_test, Y_train, Y_test = train_test_split(X_pinguin_data, Y_pinguin_data, train_size=0.85)
```

## Pinguïns Classificeren - Stap 2: het netwerk maken en trainen

> Slide: Pinguïns Classificeren - Stap 2

```python
from sklearn.neural_network import MLPClassifier

# maak een neuraal netwerk met 2 verborgen layers van elk 15 neuronen
clf = MLPClassifier(solver='sgd', max_iter=50, alpha=1e-5, learning_rate_init=0.01, hidden_layer_sizes=(15,15))

# train het model op de train set
clf.fit(X_train, Y_train)
# laat het model de soort voorspellen van de pinguïns in de test set
Y_predicted = clf.predict(X_test)
```

## Pinguïns Classificeren - Stap 2: de scores afdrukken

> Slide: Pinguïns Classificeren - Stap 2

```python
train_score = clf.score(X_train,Y_train)
test_score = clf.score(X_test,Y_test)
print("train: " + str(round(train_score*100)) + "%")
print("test: " + str(round(test_score*100)) + "%")
```
